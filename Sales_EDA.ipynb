## 📊 Sales by Region

```python
sns.barplot(x="Region", y="Sales", hue="Region", data=df, palette="Set2", legend=False)
plt.title("Sales by Region")
plt.show()
```

Insight: The South and East regions generate the highest sales, with almost equal performance, while other regions contribute comparatively less.

## 📊 Sales by Category

```python
sns.barplot(x="Category", y="Sales", hue="Category", data=df, palette="Set3", legend=False)
plt.title("Sales by Category")
plt.show()
```

Insight:
The Technology category shows higher sales compared to other categories.

## 📊 Sales by Segment

```python
sns.barplot(x="Segment", y="Sales", hue="Segment", data=df, palette="pastel", legend=False)
plt.title("Sales by Segment")
plt.show()
```

Insight: The Home Office segment generates the highest sales, while the Consumer segment has the lowest contribution.

## 📊 Top Cities

```python
plt.figure(figsize=(10,5))

top_cities = df.groupby("City")["Sales"].sum().sort_values(ascending=False).head(10)

sns.barplot(x=top_cities.index, y=top_cities.values, hue=top_cities.index, palette="Set3", legend=False)
plt.title("Top 10 Cities by Sales")
plt.xticks(rotation=45)
plt.show()
```

Insight:
New York City has the highest sales among all cities, making it the top-performing location.

## 📌 Conclusion

- Sales are highest in specific regions like West.
- Technology category dominates overall sales.
- Consumer segment contributes the most revenue.
- A few cities generate the majority of total sales.

This analysis helps businesses focus on high-performing areas and improve low-performing regions.
